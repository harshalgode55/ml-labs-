```python
numerical_cols = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
]
```

```python
# Countplot

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Type", hue="Machine failure")

plt.title("Machine Failure Count by Type")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()
```

```python
# Histogram

fig, axes = plt.subplots(2, 3, figsize=(16, 8))

for ax, col in zip(axes.flatten(), numerical_cols):
    sns.histplot(data=df, x=col, bins=30, kde=True, ax=ax)
    ax.set_title(col)

axes.flatten()[-1].axis("off")

fig.suptitle("Histogram of Numerical Features")

plt.tight_layout()
plt.show()
```

```python
# Boxplot

plot_df = df.copy()

plot_df["Machine failure"] = plot_df["Machine failure"].map({0: "No Failure", 1: "Failure"})

fig, axes = plt.subplots(2, 3, figsize=(16, 8))

for ax, col in zip(axes.flatten(), numerical_cols):
    sns.boxplot(data=plot_df, x="Machine failure", y=col, ax=ax)
    ax.set_title(col)

axes.flatten()[-1].axis("off")

fig.suptitle("Boxplot of Numerical Features by Machine Failure")

plt.tight_layout()
plt.show()
```

```python
# Heatmap

plt.figure(figsize=(10, 8))

sns.heatmap(df[numerical_cols + ["Machine failure"]].corr(),
            annot=True, fmt=".2f", cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
X = df[
    [
        "Type",
        "Air temperature [K]",
        "Process temperature [K]",
        "Rotational speed [rpm]",
        "Torque [Nm]",
        "Tool wear [min]"
    ]
]
```

```python
y = df["Machine failure"]
```

```python
print("Input shape:", X.shape)
print("Output shape:", y.shape)
```

```python
categorical_features = ["Type"]

numerical_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num",StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"),categorical_features)
    ]
)
```

## Kernel-Linear

```python
linear_svm = Pipeline(
    steps=[("preprocessor", preprocessor), ("classifier", SVC(kernel="linear"))]
)

linear_svm.fit(X_train, y_train)
```

```python
y_pred_linear = linear_svm.predict(X_test)
```

```python
accuracy_linear = accuracy_score(y_test, y_pred_linear)

precision_linear = precision_score(y_test, y_pred_linear, zero_division=0)

recall_linear = recall_score(y_test, y_pred_linear, zero_division=0)

f1_linear = f1_score(y_test, y_pred_linear, zero_division=0)

print("Accuracy :", accuracy_linear)
print("Precision:", precision_linear)
print("Recall   :", recall_linear)
print("F1 Score :", f1_linear)
```

```python
print(
    classification_report(y_test, y_pred_linear, target_names=["No Failure", "Failure"],
    zero_division = 0)
)
```

```python
cm_linear = confusion_matrix(y_test, y_pred_linear)

plt.figure(figsize=(6, 5))

sns.heatmap(cm_linear, annot=True, fmt="d", cmap="summer",
            xticklabels=["No Failure", "Failure"],
            yticklabels=["No Failure", "Failure"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Linear SVM")

plt.show()
```

```python
from sklearn.pipeline import Pipeline
pipeline = Pipeline(steps=[("preprocessor", preprocessor), ("classifier", SVC())])
```

```python
param_grid = [{"classifier__kernel": ["linear"], "classifier__C": [0.1, 1, 10, 100]},
             {"classifier__kernel": ["rbf"], "classifier__C": [0.1, 1, 10, 100], "classifier__gamma": ["scale", 0.01, 0.1, 1]}
             ]
```

```python
grid_search = GridSearchCV(estimator=pipeline, param_grid=param_grid, cv=5, scoring="f1", n_jobs=-1, verbose=1)
```

```python
grid_search.fit(X_train, y_train)

print("GridSearchCV completed.")
```

```python
print("Best Cross-Validation F1 Score:", grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_

y_pred_tuned = best_model.predict(X_test)
```

```python
accuracy_tuned = accuracy_score(y_test, y_pred_tuned)

precision_tuned = precision_score(y_test, y_pred_tuned, zero_division=0)

recall_tuned = recall_score(y_test, y_pred_tuned, zero_division=0)

f1_tuned = f1_score(y_test, y_pred_tuned, zero_division=0)

print("Tuned SVM Results")

print("Accuracy :", accuracy_tuned)
print("Precision:", precision_tuned)
print("Recall   :", recall_tuned)
print("F1 Score :", f1_tuned)
```
